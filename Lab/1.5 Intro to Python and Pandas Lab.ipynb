Dataset: Social Media Impact on Life
Source: Kaggle (Student Mental Health and Burnout Dataset)

Features: Key features include Age, Gender, Social Media Platform, Time Spent (hours), and various impact scores (Sleep, Productivity, Happiness).

The Time Sink: Which social media platform has the highest average daily usage time among students?

The Gender Gap: Is there a significant difference in reported "Productivity Loss" scores between male and female students?

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Load the dataset (Assuming the file is in the same folder as your notebook)
# You likely won't need the 'ISO' encoding for this one, standard utf-8 usually works
df = pd.read_csv('Social_media_impact_on_life.csv')

# Let's take a quick look to make sure it loaded correctly
print("Data Shape:", df.shape)
df.head()
```

```python
# Check for missing values in each column
df.isnull().sum()
```

# 1. The Time Sink: Which social media platform has the highest average daily usage time among students?

```python
# Question 1: Which platform has the highest average daily usage?
avg_usage = df.groupby('Most_Used_Platform')['Avg_Daily_Usage_Hours'].mean().sort_values(ascending=False)
print(avg_usage)
```

Takeaway: People value communication, but they also prefer a wide variety of formats and platforms-- Tiktok delivers read only content while snapchat allows back and forth in the same small bite format. The fact that WhatsApp dominates shows that direct messaging is what people value the most, presumably young teens talking to friends or their crush.

# 2. Between TikTok and Snapchat (the "bite-sized" formats), which one actually commands more time from the students?

```python
# Code
bite_sized = df[df['Most_Used_Platform'].isin(['TikTok', 'Snapchat'])]
print(bite_sized.groupby('Most_Used_Platform')['Avg_Daily_Usage_Hours'].mean())
```

Takeaway: The margin of difference between the apps is practically negligible, but reflects the simple nature of wanting immediate gratification style communication tools. Snapchat enables the back and forth while TikTok enables broadcasting soapbox style.

# 3. If people value communication: does the platform used change whether they report a "Yes" or "No" for Affects_Academic_Performance?

```python
# Code
pd.crosstab(df['Most_Used_Platform'], df['Affects_Academic_Performance'], normalize='index')
```

Takeaway: The data shows a surprising twist as a 1995 baby: Facebook actually averages HIGHER for no regarding the impact on academics! LinkedIn dominates in terms of the professional communication environment not being an attractive way to waste time, with Facebook in a VERY close second. While some applications lean entirely one way or the other, I have to imagine that there is a distinct lack of data for those applications, contrary to the more complex ratios in the other apps. Low and behold, TikTok and Snapchat are HUGELY distracting, as is Instagram!

# 4. If WhatsApp/DM is about staying up late talking to friends, is there a measurable dip in Sleep_Hours_Per_Night for WhatsApp users compared to others?

```python
# Code
df.groupby('Most_Used_Platform')['Sleep_Hours_Per_Night'].mean().sort_values()
```

Takeaway: For a healthy 8 hours of sleep, low stimulation applications like LINE and VKontakte have the smallest impact on lost sleep time; but even going just an hour and a half down the list (A mean of 6.5 hours of sleep) Youtube is JUST on the verge of being too distracting for a reliable night of sleep. TikTok isn't too far off from Youtube either, but there is a pretty major difference between LinkedIn and Twitter, showing that socials, especially the more immediately gratifying tools like Snapchat, TikTok, and WhatsApp for youth communication tools.

# 5. Is there a specific age group or "Academic Level" where these high-communication platforms dominate?

```python
# Code
df.groupby('Academic_Level')['Most_Used_Platform'].agg(pd.Series.mode)
```
